```python
import re
import pandas as pd

def extract_characters(comment, character_dict):

    # Convert comment to lowercase
    comment = str(comment).lower()

    found_characters = set()

    for character, aliases in character_dict.items():

        for alias in aliases:

            # lowercase alias
            alias = alias.lower()

            # escape special regex characters
            pattern = r"\b" + re.escape(alias) + r"\b"

            # check whether alias exists
            if re.search(pattern, comment):
                found_characters.add(character)
                break   # move to next character

    return ", ".join(sorted(found_characters))


df["Characters"] = df["clean_comment"].apply(
    lambda x: extract_characters(x, character_dict)
)
```

```python
# Create 0/1 matrix
basket = df["Characters"].str.get_dummies(sep=", ")

# Give each row a comment ID
basket.index = range(1, len(basket) + 1)
basket.index.name = "Comment_ID"

# View the result
basket.head()
```

```python
from mlxtend.frequent_patterns import apriori, association_rules
```

```python
frequent_itemsets = apriori(
    basket,
    min_support=0.005,
    use_colnames=True
)

frequent_itemsets
```

```python
frequent_itemsets["Comment_Count"] = (
    frequent_itemsets["support"] * len(basket)
).round().astype(int)

frequent_itemsets
```

```python
from mlxtend.frequent_patterns import apriori

# ==========================================
# 1. Find ONLY 2-character combinations
# ==========================================

pair_itemsets = apriori(
    basket,
    min_support=0.005,
    use_colnames=True,
    max_len=2
)

# Keep only pairs
pair_itemsets = pair_itemsets[
    pair_itemsets["itemsets"].apply(len) == 2
].copy()


# ==========================================
# 2. Calculate number of comments
# ==========================================

pair_itemsets["Comment_Count"] = (
    pair_itemsets["support"] * len(basket)
).round().astype(int)


# ==========================================
# 3. Remove very rare pairs
#    Keep pairs appearing in >= 50 comments
# ==========================================

pair_itemsets = pair_itemsets[
    pair_itemsets["Comment_Count"] >= 50
].copy()


# ==========================================
# 4. Calculate LIFT manually
# ==========================================

def calculate_lift(itemset):

    characters = list(itemset)

    char1 = characters[0]
    char2 = characters[1]

    # Probability of character 1
    p_a = basket[char1].mean()

    # Probability of character 2
    p_b = basket[char2].mean()

    # Probability of both characters
    p_ab = (
        basket[char1].astype(bool) &
        basket[char2].astype(bool)
    ).mean()

    # Lift
    lift = p_ab / (p_a * p_b)

    return lift


pair_itemsets["lift"] = pair_itemsets["itemsets"].apply(
    calculate_lift
)


# ==========================================
# 5. Create readable character names
# ==========================================

pair_itemsets["Character_1"] = pair_itemsets[
    "itemsets"
].apply(lambda x: sorted(x)[0])

pair_itemsets["Character_2"] = pair_itemsets[
    "itemsets"
].apply(lambda x: sorted(x)[1])


# ==========================================
# 6. Final table
# ==========================================

pair_results = pair_itemsets[
    [
        "Character_1",
        "Character_2",
        "Comment_Count",
        "support",
        "lift"
    ]
].sort_values(
    "lift",
    ascending=False
)


pair_results.reset_index(drop=True, inplace=True)

pair_results
```
