```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
df1=pd.read_csv("/content/all_comments.csv")
df1.head()
```

```python
df2=pd.read_csv("/content/all_comments2.csv")
df3=pd.read_csv("/content/all_comments3.csv")
df4=pd.read_csv("/content/all_comments4.csv")
df5=pd.read_csv("/content/all_comments5.csv")
```

```python
#for df in [df2, df3,df4]:
#    if 'Comment' in df.columns:
#        df.rename(columns={'Comment': 'comment'}, inplace=True)

comment_series = [
    df1['comment'],
    df2['comment'],
    df3['comment'],
    df4['comment'],
    df5['comment']
]

all_comments_df = pd.DataFrame(pd.concat(comment_series, ignore_index=True), columns=['comment'])

display(all_comments_df.head())
```

```python
all_comments_df.info()
```

```python
all_comments_df.dropna(subset=['comment'], inplace=True)
all_comments_df.info()
```

```python
all_comments_df.to_csv('all_commentsallseason.csv', index=False)
```
